```python
merged_data_set['Genres'].fillna('Unknown',inplace=True)
```

```python
merged_data_set.to_csv("merged_data_set.csv",index=False)
```

```python
#Q21  Which game genres generate the most global sales?

genre_sales = (merged_data_set.groupby('Genre')['Global_Sales'].sum().reset_index().sort_values('Global_Sales', ascending=False))

print(genre_sales.head(10))

plt.figure(figsize=(10,6))
sns.barplot(data=genre_sales, x='Global_Sales', y='Genre', palette='magma')
plt.title('Top Game Genres by Global Sales', fontsize=14, weight='bold')
plt.xlabel('Total Global Sales (in millions)', fontsize=12)
plt.ylabel('Genre', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
#Q-22 How does user rating affect global sales?

rating_sales = merged_data_set[['Rating', 'Global_Sales']].dropna()
rating_sales['Rating'] = pd.to_numeric(rating_sales['Rating'], errors='coerce')
rating_sales = rating_sales.dropna()


avg_sales_by_rating = ( rating_sales.groupby('Rating')['Global_Sales'].mean().reset_index().sort_values('Rating'))

# Plot the result
plt.figure(figsize=(8,5))
sns.lineplot(data=avg_sales_by_rating, x='Rating', y='Global_Sales', marker='o', color='orange')
plt.title('Average Global Sales by User Rating')
plt.xlabel('User Rating')
plt.ylabel('Average Global Sales (in millions)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
#  Q-23 Which platforms have the most games with high ratings (e.g., above 4)?

df = merged_data_set[['Platform', 'Rating']].dropna().copy()
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')
df = df.dropna(subset=['Rating'])

high_rated = df[df['Rating'] > 4]

top_platforms = (high_rated.groupby('Platform').size().reset_index(name='High_Rated_Game_Count').sort_values('High_Rated_Game_Count', ascending=False))

print(top_platforms.head(15))

plt.figure(figsize=(10,6))
sns.barplot(data=top_platforms.head(10), x='High_Rated_Game_Count', y='Platform', palette='viridis')
plt.title('Top Platforms with High-Rated Games (Rating > 4)', fontsize=14, weight='bold')
plt.xlabel('Number of Games (Rating > 4)', fontsize=12)
plt.ylabel('Platform', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q-24 What’s the trend of releases and sales over time?

release_sales = merged_data_set[['Year', 'Global_Sales']].dropna()

yearly_trends = (    release_sales.groupby('Year').agg(Games_Released=('Global_Sales', 'count'),Total_Global_Sales=('Global_Sales', 'sum')).reset_index().sort_values('Year'))

fig, ax1 = plt.subplots(figsize=(12,6))

# Line 1 – Number of games released
sns.lineplot(data=yearly_trends, x='Year', y='Games_Released', color='orange', label='Games Released', ax=ax1)
ax1.set_xlabel('Year')
ax1.set_ylabel('Number of Games Released', color='orange')
ax1.tick_params(axis='y', labelcolor='orange')

# Create a twin axis for sales
ax2 = ax1.twinx()
sns.lineplot(data=yearly_trends, x='Year', y='Total_Global_Sales', color='blue', label='Total Sales', ax=ax2)
ax2.set_ylabel('Total Global Sales (in millions)', color='blue')
ax2.tick_params(axis='y', labelcolor='blue')

plt.title('Trend of Game Releases and Global Sales Over Time', fontsize=14, weight='bold')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q-25  Do highly wishlisted games lead to more sales?

data = merged_data_set[['Wishlist', 'Global_Sales']].dropna().copy()
data['Wishlist'] = pd.to_numeric(data['Wishlist'], errors='coerce')
data['Global_Sales'] = pd.to_numeric(data['Global_Sales'], errors='coerce')
data = data.dropna()

data['Wishlist_Level'] = pd.qcut(data['Wishlist'], q=3, labels=['Low', 'Medium', 'High'])

wishlist_sales = (data.groupby('Wishlist_Level')['Global_Sales'].mean().reset_index().sort_values('Global_Sales', ascending=False))
print(wishlist_sales)

plt.figure(figsize=(8,5))
sns.barplot(data=wishlist_sales, x='Wishlist_Level', y='Global_Sales', palette='viridis')
plt.title('Average Global Sales by Wishlist Level')
plt.xlabel('Wishlist Level')
plt.ylabel('Average Global Sales (in millions)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()

```

```python
# Q-26 Which genres have the highest engagement but lowest sales?

data = merged_data_set[['Genres', 'Plays', 'Global_Sales']].dropna().copy()

genre_engagement = (data.groupby('Genres')[['Plays', 'Global_Sales']].mean().reset_index())

genre_engagement = genre_engagement.sort_values(by=['Plays', 'Global_Sales'], ascending=[False, True])

print(genre_engagement.head(10))

plt.figure(figsize=(10,6))
sns.scatterplot(data=genre_engagement, x='Global_Sales', y='Plays', color='teal', s=80)
plt.title('Genre Engagement vs Sales (Simplified)', fontsize=14, weight='bold')
plt.xlabel('Average Global Sales (in millions)')
plt.ylabel('Average Plays (Engagement)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

```

```python
# Q-27 Do highly listed games (wishlist/backlogs) correlate with better ratings?

data = merged_data_set[['Wishlist', 'Backlogs', 'Rating']].dropna().copy()

data['Wishlist'] = pd.to_numeric(data['Wishlist'], errors='coerce')
data['Backlogs'] = pd.to_numeric(data['Backlogs'], errors='coerce')
data['Rating'] = pd.to_numeric(data['Rating'], errors='coerce')
data = data.dropna()

data['Total_Listed'] = data['Wishlist'] + data['Backlogs']

data['Listing_Level'] = pd.qcut(data['Total_Listed'], q=3, labels=['Low', 'Medium', 'High'])

rating_trend = (data.groupby('Listing_Level')['Rating'].mean().reset_index().sort_values('Rating', ascending=False))

print(rating_trend)

plt.figure(figsize=(8,5))
sns.barplot(data=rating_trend, x='Listing_Level', y='Rating', palette='viridis')
plt.title('Average Rating by Total Listings (Wishlist + Backlogs)')
plt.xlabel('Listing Level')
plt.ylabel('Average Rating')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q-28 How does user engagement differ across genres?
data = merged_data_set[['Genres', 'Plays', 'Playing', 'Backlogs']].dropna().copy()

data['Main_Genre'] = data['Genres'].str.split(',').str[0].str.strip()

genre_engagement = (data.groupby('Main_Genre')[['Plays', 'Playing', 'Backlogs']].mean().reset_index().sort_values('Plays', ascending=False))

print(genre_engagement.head(10))

plt.figure(figsize=(12,6))
sns.barplot(data=genre_engagement.head(10), x='Main_Genre', y='Plays', palette='viridis')
plt.title('Average User Engagement (Plays) by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Plays')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q-29 What are the top-performing combinations of Genre + Platform?

data = merged_data_set[['Genre', 'Platform', 'Global_Sales']].dropna().copy()

combo_performance = (data.groupby(['Genre', 'Platform'])['Global_Sales'].sum().reset_index().sort_values(by='Global_Sales', ascending=False))

print(combo_performance.head(10))  

plt.figure(figsize=(12,6))
sns.barplot(data=combo_performance.head(10),x='Global_Sales',y='Genre',hue='Platform',palette='magma')
plt.title('Top Performing Genre + Platform Combinations (by Global Sales)')
plt.xlabel('Total Global Sales (in millions)')
plt.ylabel('Genre')
plt.legend(title='Platform', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q-30 What does a regional sales heatmap by genre reveal?

data = merged_data_set[['Genre', 'NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].dropna().copy()

region_sales = (data.groupby('Genre')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum().reset_index())

region_sales.set_index('Genre', inplace=True)

plt.figure(figsize=(10,6))
sns.heatmap(region_sales, annot=True, fmt=".1f", cmap='viridis')
plt.title('Regional Sales by Genre (in millions)', fontsize=14, weight='bold')
plt.xlabel('Region')
plt.ylabel('Genre')
plt.tight_layout()
plt.show()
```
