```python
# Q 10 Which region generates the most game sales?


total_sales_by_region = {'NA': sales['NA_Sales'].sum(),'EU': sales['EU_Sales'].sum(),'JP': sales['JP_Sales'].sum(),'Other': sales['Other_Sales'].sum()}


region_sales = pd.DataFrame({
    'Region': ['NA','EU','JP','Other'],
    'Total_Sales': [sales['NA_Sales'].sum(),sales['EU_Sales'].sum(),sales['JP_Sales'].sum(),sales['Other_Sales'].sum()]}).sort_values(by='Total_Sales', ascending=False)
print(region_sales)

plt.figure(figsize=(8,5))
plt.bar(region_sales['Region'], region_sales['Total_Sales'], color=['blue', 'orange', 'green', 'red'])
plt.title('Total Game Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales (in millions)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q11  What are the best-selling platforms?

best_platforms = sales.groupby('Platform')['Global_Sales'].sum().reset_index()
best_platforms = best_platforms.sort_values(by='Global_Sales', ascending=False)
best_platforms.head(10)

plt.figure(figsize=(10,6))
plt.barh(best_platforms['Platform'].head(10), best_platforms['Global_Sales'].head(10))
plt.gca().invert_yaxis()
plt.title('Top 10 Best-Selling Platforms')
plt.xlabel('Total Global Sales (in millions)')
plt.ylabel('Platform')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q12  What’s the trend of game releases and sales over years?
sales['Year'] = pd.to_numeric(sales['Year'], errors='coerce')                       
sales['Year'] = sales['Year'].astype(int) 

yearly_trends = sales.groupby('Year').agg(Total_Games=('Name', 'count'),Total_Sales=('Global_Sales', 'sum')).reset_index()
print(yearly_trends)

plt.figure(figsize=(10,6))
plt.plot(yearly_trends['Year'], yearly_trends['Total_Games'], marker='o', color='skyblue')
plt.title('Trend of Game Releases Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Games Released')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10,6))
plt.plot(yearly_trends['Year'], yearly_trends['Total_Sales'], marker='s', color='orange')
plt.title('Trend of Global Game Sales Over the Years')
plt.xlabel('Year')
plt.ylabel('Total Sales (in millions)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q13 Who are the top publishers by sales?
top_publishers = (sales.groupby('Publisher')['Global_Sales'].sum().reset_index().sort_values(by='Global_Sales', ascending=False)).head(10)
print(top_publishers)

plt.figure(figsize=(10,6))
plt.barh(top_publishers['Publisher'].head(10), top_publishers['Global_Sales'].head(10))
plt.gca().invert_yaxis()
plt.title('Top 10 Publishers by Global Sales')
plt.xlabel('Total Global Sales (in millions)')
plt.ylabel('Publisher')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
#  Q 14 Which games are the top 10 best-sellers globally?
top_games = sales.sort_values(by='Global_Sales', ascending=False).head(10)
print(top_games[['Name', 'Platform', 'Year', 'Publisher', 'Global_Sales']])

plt.figure(figsize=(10,6))
plt.barh(top_games['Name'], top_games['Global_Sales'], color='orange')
plt.gca().invert_yaxis()  # highest at top
plt.title('Top 10 Best-Selling Games Globally')
plt.xlabel('Global Sales (in millions)')
plt.ylabel('Game Title')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
#Q-15  How do regional sales compare for specific platforms?

regional_sales = (sales.groupby('Platform')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum().reset_index())
print(regional_sales.head(10))

plt.figure(figsize=(12,6))
regional_sales.set_index('Platform').head(10).plot(kind='bar', stacked=True, figsize=(12,6))
plt.title('Regional Sales Comparison Across Top 10 Platforms')
plt.xlabel('Platform')
plt.ylabel('Total Sales (in millions)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

```

```python
#Q16 How has the market evolved by platform over time?

platform_trends = (sales.groupby(['Year', 'Platform'])['Global_Sales'].sum().reset_index())


platform_trends = platform_trends.sort_values(by='Year')

print(platform_trends)

top_platforms = (sales.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False).head(5).index)

plt.figure(figsize=(12,7))
for platform in top_platforms:
    platform_data = platform_trends[platform_trends['Platform'] == platform]
    plt.plot(platform_data['Year'], platform_data['Global_Sales'], marker='o', label=platform)

plt.title('Evolution of Global Game Sales by Platform Over Time')
plt.xlabel('Year')
plt.ylabel('Global Sales (in millions)')
plt.legend(title='Platform', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
#  Q-17 What are the regional genre preferences?

genre_sales = (sales.groupby('Genre')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum().reset_index().sort_values(by='NA_Sales', ascending=False))

print(genre_sales)

genre_sales.set_index('Genre').plot(kind='bar', figsize=(12,6),stacked=True)

plt.title('Regional Sales by Game Genre')
plt.xlabel('Genre')
plt.ylabel('Total Sales (in millions)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

```

```python
#Q-18  What’s the yearly sales change per region?

yearly_region_sales = (sales.groupby('Year')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum().reset_index().sort_values('Year'))

plt.figure(figsize=(12,6))

regions = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']
for region in regions:
    plt.plot(
        yearly_region_sales['Year'],
        yearly_region_sales[region],
        marker='o',
        linewidth=2,
        label=region.replace('_Sales', '')
    )

plt.title('Yearly Sales Trend by Region', fontsize=14, weight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Total Sales (in millions)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.legend(title='Region', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()


```

```python
# Q19 What is the average sales per publisher?

avg_sales_per_publisher = (sales.groupby('Publisher')['Global_Sales'].mean().reset_index().sort_values('Global_Sales', ascending=False))

top_avg_publishers = avg_sales_per_publisher.head(10)
print(top_avg_publishers)

plt.figure(figsize=(10,6))
plt.barh(top_avg_publishers['Publisher'], top_avg_publishers['Global_Sales'], color='orange')
plt.gca().invert_yaxis()  # So the highest is at the top
plt.title('Top 10 Publishers by Average Global Sales', fontsize=14, weight='bold')
plt.xlabel('Average Global Sales (in millions)', fontsize=12)
plt.ylabel('Publisher', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
#Q-20 What are the top 5 best-selling games per platform?

sales_sorted = sales.sort_values(['Platform', 'Global_Sales'], ascending=[True, False])

top5_games_per_platform = sales_sorted.groupby('Platform').head(5).reset_index(drop=True)

print(top5_games_per_platform[['Platform', 'Name', 'Global_Sales']].head(20))


sample_platforms = top5_games_per_platform[top5_games_per_platform['Platform'].isin(['PS2', 'Wii', 'X360'])]

plt.figure(figsize=(12,6))
sns.barplot(
    data=sample_platforms,
    x='Global_Sales',
    y='Name',
    hue='Platform'
)
plt.title('Top 5 Best-Selling Games per Selected Platform', fontsize=14, weight='bold')
plt.xlabel('Global Sales (in millions)', fontsize=12)
plt.ylabel('Game', fontsize=12)
plt.legend(title='Platform')
plt.tight_layout()
plt.show()
```
