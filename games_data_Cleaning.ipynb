```python
import numpy as np

def convert_k_to_numeric(value):
    if pd.isna(value):
        return np.nan   # ✅ keeps NaN type consistent for numeric columns
    s = str(value).strip().upper()
    if s.endswith('K'):
        return float(s[:-1]) * 1000
    else:
        return float(s)
```

```python
games['Plays'] = games['Plays'].apply(convert_k_to_numeric)
```

```python
games_cleaned = pd.read_csv('games_data_cleaned.csv')
```

```python
# Question 1 -- >  What are the top-rated games by user reviews?
Q1=games_cleaned.sort_values(by='Rating',ascending=False).head(10)
print(Q1[['Rating','Name']])

ax = Q1.plot.barh(
    x='Name',y='Rating',figsize=(10, 5),color='green',title='Top 10 Highest Rated Games by User Reviews')

# Fix the reversed order
ax.invert_yaxis()

ax.set_xlabel('Rating')
ax.set_ylabel('Game Name')
ax.grid(True, axis='x', linestyle='--', alpha=0.6)

```

```python
# Question 2 --> Which developers (teams) have the highest average ratings?

Q2 = (games_cleaned.groupby('Team')['Rating'].mean().sort_values(ascending=False).head(10))

print(Q2)

plt.figure(figsize=(10, 6))
Q2.plot(kind='bar', color='green')
plt.title('Top 10 Developer Teams by Average Rating')
plt.xlabel('Developer / Team')
plt.ylabel('Average Rating')
plt.xticks(rotation=45, ha='right')
plt.grid(True, axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q3 -->  What are the most common genres in the dataset?
games_cleaned = games_cleaned.dropna(subset=['Genres'])

games_cleaned['Genres'] = games_cleaned['Genres'].str.split(',')
games_cleaned = games_cleaned.explode('Genres')

genre_counts = games_cleaned['Genres'].value_counts()
print(genre_counts)

games_cleaned['Genres'] = games_cleaned['Genres'].str.strip()

plt.figure(figsize=(10,6))
genre_counts.plot(kind='bar', edgecolor='black')
plt.title('Most Common Game Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Games')
plt.tight_layout()
plt.show()
```

```python
games_cleaned.head(1)
```

```python
# Q4 --> Which games have the highest backlog compared to wishlist?
games_cleaned['Backlog_Wishlist_Diff'] = games_cleaned['Backlogs'] - games_cleaned['Wishlist']
sorted_diff = games_cleaned.sort_values(by='Backlog_Wishlist_Diff', ascending=False).head(10)
print(sorted_diff[['Name', 'Backlogs', 'Wishlist', 'Backlog_Wishlist_Diff']])

plt.figure(figsize=(10, 5))
plt.barh(sorted_diff['Name'], sorted_diff['Backlog_Wishlist_Diff'])
plt.xlabel('Backlog - Wishlist (absolute difference)')
plt.ylabel('Game Name')
plt.title('Top 10 Games with Highest Backlog Compared to Wishlist')
plt.gca().invert_yaxis()  
plt.tight_layout()
plt.show()
```

```python
#  Q5  What is the game release trend across years?
games_cleaned['Release Date'] = pd.to_datetime(games_cleaned['Release Date'], errors='coerce')
games_cleaned = games_cleaned.dropna(subset=['Release Date'])
games_cleaned['Year'] = pd.to_datetime(games_cleaned['Release Date']).dt.year
games_per_year = games_cleaned.groupby('Year')['Name'].count().reset_index()
games_per_year.rename(columns={'Name': 'Game_Count'}, inplace=True)
print(games_per_year)

plt.figure(figsize=(10,6))
plt.plot(games_per_year['Year'], games_per_year['Game_Count'], marker='o')
plt.title('Game Release Trend Across Years')
plt.xlabel('Year')
plt.ylabel('Number of Games Released')
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
# Q6 What is the distribution of user ratings?

print(games_cleaned['Rating'].describe())

plt.figure(figsize=(10,6))
plt.hist(games_cleaned['Rating'], bins=5, range=(0,5), edgecolor='black')
plt.title('Distribution of User Ratings (0–5 Scale)')
plt.xlabel('User Rating')
plt.ylabel('Number of Games')
plt.grid(True, linestyle='--', alpha=1)
plt.tight_layout()
plt.show()
```

```python
#Q7 What are the top 10 most wishlisted games?
unique_games = games_cleaned.drop_duplicates(subset='Name')
top10_wishlisted = unique_games.sort_values(by='Wishlist',ascending=False).head(10)
print(top10_wishlisted[['Name', 'Wishlist']])

plt.figure(figsize=(10,6))
plt.barh(top10_wishlisted['Name'], top10_wishlisted['Wishlist'])
plt.gca().invert_yaxis()  # highest on top
plt.title('Top 10 Most Wishlisted Games')
plt.xlabel('Number of Wishlists')
plt.ylabel('Game Name')
plt.tight_layout()
plt.show()

```

```python
# Q8 What’s the average number of plays per genre?

avg_plays = games_cleaned.groupby('Genres')['Plays'].mean().reset_index()
print(avg_plays)

plt.figure(figsize=(10,6))
plt.barh(avg_plays['Genres'], avg_plays['Plays'])
plt.gca().invert_yaxis()  
plt.title('Average Number of Plays per Genre')
plt.xlabel('Average Plays')
plt.ylabel('Genre')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Q9 Which developer studios are the most productive?
studio_summary = games_cleaned.groupby('Team').agg(Total_Games=('Name', 'count'), Avg_Rating=('Rating', 'mean'),Total_Plays=('Plays', 'sum'),Total_Wishlist=('Wishlist', 'sum')).reset_index()

most_productive = studio_summary.sort_values(by='Total_Games', ascending=False).head(10)
print(most_productive[['Team', 'Total_Games']])

plt.figure(figsize=(10,6))
plt.barh(most_productive['Team'], most_productive['Total_Games'])
plt.gca().invert_yaxis()
plt.title('Top 10 Most Productive Developer Studios')
plt.xlabel('Number of Games')
plt.ylabel('Developer Studio')
plt.tight_layout()
plt.show()

```

```python
import pandas as pd

path = r"d:\Guvi_Practice\Video_Game_project\games_data_cleaned.csv"

# Load
df = pd.read_csv(path)

# Clean TBD (make it NaT)
df['Release Date'] = df['Release Date'].astype(str).str.strip()
df.loc[df['Release Date'].str.lower().str.contains('tbd', na=False), 'Release Date'] = pd.NaT

# Convert to datetime
df['Release Date'] = pd.to_datetime(df['Release Date'], errors='coerce')

# Fill NaT with median date
median_date = df['Release Date'].median()
df['Release Date'] = df['Release Date'].fillna(median_date)

# Save back
df.to_csv(path, index=False, date_format='%Y-%m-%d')

print("Done da! Filled TBD rows with median date 👍")
```
